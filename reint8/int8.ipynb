```python
import struct

def u23_to_f32(x: int) -> float:
    """Convert an integer in range [0, 2^23) to a float exactly.
    
    Produces an incorrect result for integers outside the range.
    """
    # Magic number: bit representation of 2^23 as float
    magic_float = float(1 << 23)
    magic_bits = struct.unpack('I', struct.pack('f', magic_float))[0]
    
    # XOR input with magic bits and convert back to float
    result_bits = x ^ magic_bits
    result_float = struct.unpack('f', struct.pack('I', result_bits))[0]
    
    return result_float - magic_float
```

```python
import struct

def i23_to_f32(x: int, scale: float = 1.0, local_max: float = 0.0) -> float:
    """Convert an integer in range [0, 2^23) to a float exactly.
    
    Produces an incorrect result for integers outside the range.
    """
    # Magic number: bit representation of 2^23 as float
    magic_float = float(1 << 23) + float(1 << 22)
    # magic_float = float(1 << 24)
    magic_bits = struct.unpack('I', struct.pack('f', magic_float))[0]
    
    # XOR input with magic bits and convert back to float
    # Mask to ensure unsigned 32-bit integer range
    # x = ~(x - 1) # convert to 1's complement
    # result_bits = (x + magic_bits) & 0xFFFFFFFF
    result_bits = (x + magic_bits)
    result_float = struct.unpack('f', struct.pack('I', result_bits))[0]
    
    return result_float * scale - ((magic_float * scale) + local_max)

def naive_i23_to_f32(x: int, scale: float = 1.0, local_max: float = 0.0) -> float:
    return x * scale - local_max

# neg_u23_to_f32(-29)
scale = 4.24
local_max = 50
res = i23_to_f32(-30001, scale, local_max)
naive_res = naive_i23_to_f32(-30001, scale, local_max)
res, naive_res, res - naive_res
```
